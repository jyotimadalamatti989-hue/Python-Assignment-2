```python
# Accepting two numbers from the user and converting them to float for mathematical operations
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Performing calculations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2

# Checking for division by zero before performing division
if num2 != 0:
    division = num1 / num2
else:
    division = "Undefined (Cannot divide by zero)"

# Displaying each result separately
print(f"\nAddition: {addition}")
print(f"Subtraction: {subtraction}")
print(f"Multiplication: {multiplication}")
print(f"Division: {division}")
```

```python
# Storing the verified static correct password
correct_password = "1234"

# Initializing an infinite conditional loop structure
while True:
    entered_password = input("Enter the password: ")

    # Comparing input string against requirements
    if entered_password == correct_password:
        print("Access Granted!")
        break  # Exits the while loop once the condition is satisfied
    else:
        print("Incorrect password. Please try again.")
```

```python
# Collecting product and pricing details from the user
product_name = input("Enter Product Name: ")
quantity = int(input("Enter Quantity: "))
price = float(input("Enter Price per unit: "))

# Calculating the core cost
subtotal = quantity * price

# Bonus Challenge: Adding 18% GST and a 5% discount
gst_rate = 0.18
discount_rate = 0.05

gst_amount = subtotal * gst_rate
discount_amount = subtotal * discount_rate
total_bill = subtotal + gst_amount - discount_amount

# Displaying a professionally structured shopping bill using f-strings
print("\n" + "="*30)
print("         SHOPPING BILL        ")
print("="*30)
print(f"Product Name : {product_name}")
print(f"Quantity     : {quantity}")
print(f"Price/Unit   : ${price:.2f}")
print("-"*30)
print(f"Subtotal     : ${subtotal:.2f}")
print(f"GST (18%)    : +${gst_amount:.2f}")
print(f"Discount (5%): -${discount_amount:.2f}")
print("-"*30)
print(f"TOTAL PAYABLE: ${total_bill:.2f}")
print("="*30)
```
